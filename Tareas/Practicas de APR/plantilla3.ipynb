```python
import numpy as np

class QApproxAgent:
    def __init__(self, n_features, n_actions, alpha=0.01, gamma=0.99, epsilon=0.1):
        self.n_features = n_features
        self.n_actions = n_actions
        self.alpha = alpha
        self.gamma = gamma
        self.epsilon = epsilon
        self.weights = np.zeros((n_actions, n_features))

    def q_values(self, state):
        return np.dot(self.weights, state)

    def choose_action(self, state):
        if np.random.rand() < self.epsilon:
            return np.random.randint(self.n_actions)
        return np.argmax(self.q_values(state))

    def update(self, state, action, reward, next_state):
        best_next = np.max(self.q_values(next_state))
        td_target = reward + self.gamma * best_next
        td_error = td_target - self.q_values(state)[action]
        self.weights[action] += self.alpha * td_error * state
```
