1️⃣ Métodos de acción-valor (Action-Value Methods)

Idea principal:

Estimar el valor esperado de cada acción 
𝑄(𝑠,𝑎)
Q(s,a).

Elegir la acción con mayor valor (explotación) o explorar acciones menos conocidas.

Cómo funciona:

Mantienes una tabla Q (estado → acción → valor).

Cada acción se evalúa por su recompensa promedio esperada.

Exploración: usar 
𝜖
ϵ-greedy (probabilidad de elegir acción aleatoria).

Explotación: elegir acción con mayor 
𝑄(𝑠,𝑎)
Q(s,a).

Implementación general:

# Inicialización
Q = defaultdict(lambda: [0 for _ in range(n_acciones)])
epsilon = 0.1

def elegir_accion(s):
    if np.random.rand() < epsilon:
        # Exploración
        return np.random.choice(n_acciones)
    else:
        # Explotación
        return np.argmax(Q[s])


Ventaja: simple y funciona para entornos discretos.
Aplicable: póker simplificado, tic-tac-toe, laberintos.

2️⃣ Implementación incremental (Incremental Implementation)

Idea principal:

Actualizar el valor de una acción paso a paso, sin almacenar todas las recompensas pasadas.

Permite manejar muchas iteraciones sin grandes estructuras de datos.

Cómo funciona:

𝑄𝑛+1=𝑄𝑛+𝛼(𝑅−𝑄𝑛)
Qn+1=Qn+α(R−Qn)
𝑄𝑛Qn = valor actual de la acción

𝑅
R = recompensa obtenida

𝛼
α = tasa de aprendizaje (learning rate)

Implementación general:

alpha = 0.1  # tasa de aprendizaje
Q[s][a] += alpha * (recompensa - Q[s][a])


Ventaja: no necesitas guardar todas las recompensas históricas.
Aplicable: entornos grandes o continuos, juegos con muchos estados.

Una vez que prueba una acción, su valor real reemplaza el valor inicial.

Aplicable: cuando quieres que el agente pruebe todas las acciones al inicio, como póker para todas las combinaciones de apuestas.

4️⃣ Selección de acciones con intervalo de confianza (Upper Confidence Bound, UCB)

Idea principal:

Balancea exploración y explotación usando estadísticas de incertidumbre.

Las acciones menos probadas tienen un bonus de incertidumbre, haciendo que se exploren.

Fórmula típica:

𝑎=argmax𝑎[𝑄(𝑠,𝑎)+𝑐⋅ln⁡𝑡𝑁(𝑠,𝑎)]
a=argamax[Q(s,a)+c⋅N(s,a)lnt]

𝑄(𝑠,𝑎)
Q(s,a) = valor promedio de acción

𝑁(𝑠,𝑎)
N(s,a) = número de veces que se ha tomado la acción

𝑡
t = número total de pasos

𝑐
c = parámetro que ajusta el nivel de exploración

Implementación general:

import numpy as np

t = sum(N_s)  # total de acciones tomadas
valores = [Q[s][a] + c * np.sqrt(np.log(t)/N[s][a]) for a in range(n_acciones)]
accion = np.argmax(valores)


Ventaja: explora inteligentemente solo donde hay incertidumbre.
Aplicable: juegos con muchas acciones posibles y entornos donde algunas acciones raras podrían ser muy buenas.

5️⃣ Algoritmos de Gradiente (Policy Gradient Methods)

Idea principal:

No estiman Q(s,a) directamente.

Aprenden una política estocástica parametrizada (por ejemplo, con una red neuronal)

La política genera probabilidades para cada acción y se ajusta según las recompensas recibidas.

Cómo funciona:

Aplicable: póker completo, robótica, juegos de estrategia complejos.
