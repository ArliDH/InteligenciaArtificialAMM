Algunas funciones de libreria random

| Función                 | Descripción                                    | Ejemplo                       |
| ----------------------- | ---------------------------------------------- | ----------------------------- |
| `random.random()`       | Genera un número decimal aleatorio entre 0 y 1 | `0.7345...`                   |
| `random.randint(a, b)`  | Genera un entero aleatorio entre `a` y `b`     | `random.randint(1, 10)` → 6   |
| `random.choice(lista)`  | Selecciona un elemento aleatorio de una lista  | `choice(["a","b","c"])` → "b" |
| `random.shuffle(lista)` | Desordena los elementos de una lista           | `[1,2,3,4] → [3,1,4,2]`       |


Algunas funciones de Libreria itertools

| Método           | Descripción                                         | Ejemplo                       |
| ---------------- | --------------------------------------------------- | ----------------------------- |
| `most_common(n)` | Retorna los `n` elementos más frecuentes            | `most_common(1)` → [(3, 3)]   |
| `elements()`     | Devuelve los elementos repetidos su número de veces | list(Counter(...).elements()) |

```python
#Combinar elementos de un array [1,2,3]
lista = [1, 2, 3]
print(list(combinations(lista, 2)))
```

```python
#Agrupar elementos que se repitan en diferentes celdas [1,2,1,2,3,4]  = [1,1], [2,2], [3], [4]
lista = [1,1,2,2,2,3,1]
for key, group in groupby(lista):
    print(key, list(group))
```

```python
#Cuenta cuantos valores de un array se repite: 
numeros = [1,2,2,3,3,3]
print(Counter(numeros))
Counter({3: 3, 2: 2, 1: 1})
```
