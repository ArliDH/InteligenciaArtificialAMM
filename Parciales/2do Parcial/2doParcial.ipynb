```python
class entorno():
    def __init__(self):
        self.cuadro=[0,1,2,3]
    
    def desordenar_puzzle(self):
        self.cuadromodificado=[]
        self.cuadromodificado.append.random.shuffle(self.cuadro)
        #self.cuadromodificado = random.shuffle(self.cuadro)
        return self.cuadromodificado
    
    def reglas_juego(self, Moves):
        seleccionar = random(1,4)
        cuadro_seleccionado = 0
        cuadro_que_cambiara = 0

        cuadro_seleccionado = self.cuadromodificado[seleccionar]
        if seleccionar == len(self.cuadro) & Moves == 1:
            return 1
        elif seleccionar == 0 & Moves == -1:
            return -1
        
        if Moves == 1:
            cuadro_que_cambiara = self.cuadromodificado[seleccionar+1]
            self.cuadromodificado[seleccionar] = cuadro_que_cambiara
            self.cuadromodificado[seleccionar+1] = cuadro_seleccionado
        elif Moves == -1:
            cuadro_que_cambiara = self.cuadromodificado[seleccionar-1]
            self.cuadromodificado[seleccionar] = cuadro_que_cambiara
            self.cuadromodificado[seleccionar-1] = cuadro_seleccionado
        
        return self.cuadromodificado
    
    def gameover(self, reward):
        for i in range(len(self.cuadro)):
            if i == 0:
                if self.cuadromodificado[i] == 0:
                    reward+=1
            elif i == 1:
                if self.cuadromodificado[i] == 1:
                    reward+=1
            elif i == 2:
                if self.cuadromodificado[i] == 2:
                    reward+=1
            elif i == 3:
                if self.cuadromodificado[i] == 3:
                    reward+=1
        
        if reward == 4:
            return True
        else:
            return False
```
